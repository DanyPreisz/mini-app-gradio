# Mini App con Gradio

Proyecto de interfaz interactiva listo para **Google Colab**.

**Pregunta:** ¿cómo crear una mini app para un modelo de ML?

**Método:** Gradio Interface

**Stack:** gradio, scikit-learn

Modelo simple de clasificación.

## 0. Setup

```python
!pip install -q gradio

import gradio as gr
import numpy as np
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
print("OK")
```

## 1. Modelo simple (Iris)

```python
iris = load_iris()
X, y = iris.data, iris.target
model = RandomForestClassifier(random_state=42)
model.fit(X, y)
print("Modelo entrenado")
```

## 2. Función de predicción

```python
def predecir(sepal_length, sepal_width, petal_length, petal_width):
    features = np.array([[sepal_length, sepal_width, petal_length, petal_width]])
    pred = model.predict(features)[0]
    return iris.target_names[pred]

print(predecir(5.1, 3.5, 1.4, 0.2))
```

## 3. Interfaz Gradio

```python
iface = gr.Interface(
    fn=predecir,
    inputs=[
        gr.Number(label="Sepal Length"),
        gr.Number(label="Sepal Width"),
        gr.Number(label="Petal Length"),
        gr.Number(label="Petal Width")
    ],
    outputs=gr.Textbox(label="Especie predicha"),
    title="Clasificador de Iris",
    description="Ingresá las medidas de la flor"
)

# iface.launch()
print("Descomentá para lanzar la app")
```

## 4. Conclusiones

- Gradio permite crear interfaces interactivas en pocas líneas.
- Ideal para demos de modelos de ML.
- Se puede extender a imágenes, texto o audio.
